## 1. wnigs 설치
### 설명
- 파이썬의 xlwings 라이브러리는 엑셀과 연계를 통해 데이터 분석 및 자동화를 신속하게 수행할 수 있는 강력한 도구.
- 엑셀 프로그램 자체를 컨트롤 한다.
- **DRM에서 자유롭다.**

### 구성요소
- App – 프로그램
- Book – 워크북
- Sheet – 워크시트
- Range – 셀 범위


```python
 #  pip install xlwings
```

```python
import xlwings as xw

# 엑셀 앱 만들기
app = xw.App(add_book=False)
app
```

```python
# 엑셀 워크북 만들기
wb = app.books.add()
wb
```

```python
ws = wb.sheets['Sheet1']
```

```python
# 시트 이름 변경
ws.name = "수배로직설계팀"
```

```python
# 다른 이름으로 저장
wb.save("팀리스트.xlsx")
```

```python
# 워크북 닫기
wb.close()
```

```python
# 엑셀 앱 닫기
app.quit()
```

```python
# 엑셀 앱 만들어서 파일 불러오기
app = xw.App(add_book=False)
wb = app.books.open('팀리스트.xlsx')
```

```python
# 새로운 시트 생성
wb.sheets.add("영업2팀")
wb.sheets.add("마케팅팀")
```

```python
wb.sheets.add(name="세무회계팀", before="영업2팀")
```

```python
wb.sheets.add("영업3팀", after="영업2팀")
```

```python
# 워크 시트 선택(인덱스)
wb.sheets[1]
```

```python
# 시트 삭제
wb.sheets['영업3팀'].delete()
```

```python
# 시트 활성화
wb.sheets["마케팅팀"].activate()
```
