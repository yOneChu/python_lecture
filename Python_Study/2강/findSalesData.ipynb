```python
import pandas
# PID 시뮬레이터 결과만
# 품번으로 속성값 조회
import requests

def pidExecute(pid, hogi, testFlag, floor, isfloor):
    """
    Spring Controller (/findProductBom) 호출 예시
    파라미터: productNo
    응답: JSON (ArrayList<ProductDto>)
    """
    # 1️⃣ API 주소
    url = "https://vault-in.hdel.co.kr:8070/subae/pidExecute"  # 실제 서버 주소로 수정

    # 2️⃣ GET 방식으로 파라미터 전달
    params = {
        "pid": pid,
        "hogi": hogi,
        "testVersion": testFlag,
        "floor": floor,
        "isfloor": isfloor
    }

    try:
        # 3️⃣ 요청 전송
        response = requests.get(url, params=params)
        response.raise_for_status()  # 오류 시 예외 발생

        # 4️⃣ 응답 데이터 (JSON → Python 리스트)
        data = response.json()
        print("✅ 데이터 조회 성공")
        return data

    except requests.exceptions.RequestException as e:
        print("❌ 요청 실패:", e)
        return None


if __name__ == "__main__":
    pid = "CAL_AA_FRAME"
    hogi = "TEST-611639"
    testFlag = ""
    floor = ""
    isfloor = ""
    result = pidExecute(pid, hogi, testFlag, floor, isfloor)

    #print(result)
    print(result.get('AA_FRAME'))
```

```python
# 시뮬돌려서 계산값 구하기

calDic = {}

for idx, row in df.iterrows():
    #df.loc[idx, 'F'] = row['C'] + row['D']
    print(idx, row['호기번호'])
    calValue = pidExecute('CAL_AA_FRAME', row['호기번호'], '', '', '')
    calDic[row['호기번호']]= calValue.get('AA_FRAME')

```
