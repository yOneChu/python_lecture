# MLB 부품 레벨 조회


```python
# 품번으로 속성값 조회
import requests

def find_PartInfo(partNo):
    """
    Spring Controller (/findProductBom) 호출 예시
    파라미터: productNo
    응답: JSON (ArrayList<ProductDto>)
    """
    # 1️⃣ API 주소
    url = "https://vault-in.hdel.co.kr:8070/mlb/findPartOneWithPartNo"  # 실제 서버 주소로 수정

    # 2️⃣ GET 방식으로 파라미터 전달
    params = {
        "partNo": partNo
    }

    try:
        # 3️⃣ 요청 전송
        response = requests.get(url, params=params)
        response.raise_for_status()  # 오류 시 예외 발생

        # 4️⃣ 응답 데이터 (JSON → Python 리스트)
        data = response.json()
        print("✅ 데이터 조회 성공")
        return data

    except requests.exceptions.RequestException as e:
        print("❌ 요청 실패:", e)
        return None


if __name__ == "__main__":
    product_no = "37510339G06HO"
    result = find_PartInfo(product_no)

    #print(result)
    print(result.get("oid"))
    print(result.get("partNo"))
    print(result.get("partName"))
    print(result.get("spec"))
```

```python
# A1부터 C5 범위를 DataFrame으로 읽기
# import pandas as pd
ws = wb.sheets['결과']

# df = ws.range('A1:K3903').options(pd.DataFrame, header=True, index=False).value
# df = ws.range('A1').options(pd.DataFrame, expand='table').value
data = ws.range('A1:K3903').value

# 데이터를 pandas로 변환
df = pd.DataFrame(data[1:], columns=data[0])
df

```

```python
# 특정 열 추출
# iloc: 인덱스 기반 선택자
# 0열의 값 모두 추출
#
cli1_list = df.iloc[:, 0].tolist()
cli1_list
```

```python
# 사이즈 맞나 확인
len(cli1_list)
```

```python
# 근데 중복이 있어서 중복을 제거하고 데이터를 담고 싶어.
# drop_duplicates()는 pandas에서 중복된 값만 제거하면서 순서 유지
cli1_list = df.iloc[:, 0].drop_duplicates().tolist()
len(cli1_list)
```

```python
# 데이터를 출력할 새로운 시트 생성

wb.sheets.add("수배로직설계팀")

ws = wb.sheets['수배로직설계팀']
```

```python
# find_product_bom
# ws.range('A2').value = ['이르지얼', 5]
# 순서(index)도 알면서 조회하고 싶다. -> enumerate()함수

dataList = []

for idx, partNo in enumerate(cli1_list, start=1):
    print(idx, partNo)

    resultVal = find_PartInfo(partNo)

    # oid = resultVal.get("oid")
    # partNo = resultVal.get("partNo")
    # partName = resultVal.get("partName")
    # version = resultVal.get("version")
    # spec = resultVal.get("spec")
    # partSize = resultVal.get("partSize")
    # blockNo = resultVal.get("blockNo")
    # nation = resultVal.get("nation")
    # status = resultVal.get("status")

    # 데이터 없는거 key-val 지우기
    partInfo = {k: v for k, v in resultVal.items() if v is not None}
    # print(partInfo)
    dataList.append(partInfo)



    indexVal = 'A' + str(idx)
    #ws.range(f"A{idx}").value = f"결과: {resultVal}"
    #ws.range(f"A{idx}").value = list(resultVal.values())
    #ws.range(indexVal).value = list(resultVal.values())

## 리스트에 담긴 dic들을 pandas에 담기
df = pd.DataFrame(dataList)

df
```

```python
# 데이터가 NaN값만 있는 행(row) 지우기
# dropna()
df = df.dropna(how='all')
df
df.head()
```
