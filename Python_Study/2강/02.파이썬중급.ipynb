## 반복문
for i in 범위:

```python
### for  기본
array = [273, 32, 103, 57, 52]
for item in array:
    print(item)
```

```python
# for 문
# 0~4까지 출력
for i in range(5):
    print(i)

print()

# 5에서 9까지 출력
for i in range(5, 10):
    print(i)

print()

# 1부터 3씩 더해서 9까지 출력
for i in range(1, 10, 3):
    print(i)

```

## 딕셔너리
- 키 값을 기반으로 여러 자료를 저장하는 자료형
- 키: 딕셔너리 내부에서 값에 접근할 때 사용하는 것
- 값: 딕셔너리 내부에 있는 각각의 내용을 의미


```python
### 딕셔너리 반복문
dictionary = {
    "name": "망고",
    "type": "당절임",
    "ingredient": ["망고", "설탕", "메타중이황산나트륨", "치자황색소"],
    "origin": "필리핀"
}

print(dictionary)
print("1 name:", dictionary["name"])
print()

# 값 변경
print("-- 값 변경")
dictionary["name"] = "수박"
print("2 name:", dictionary["name"])

print()
# 값 추가
print("-- 값 추가")
dictionary["price"] = 100
dictionary
```

```python
### 값 제거
dictionary = {
    "name": "망고",
    "type": "당절임",
    "origin": "필리핀"
}

# 값 제거
del dictionary["name"]
print(dictionary)

```

```python
### dictionary 내부 키 확이
dictionary = {
    "name": "망고",
    "type": "당절임",
    "origin": "필리핀"
}

# 키 입력받기
key = input("> 키 입력하세요: ")

# 출력
if key in dictionary:
    print(dictionary[key])
else:
    print("해당 키 값은 존재하지 않습니다.")
```

```python
### dictionary 반복문
dictionary = {
    "name": "baseball",
    "hobby": "football",
    "address": "korea"
}

for key in dictionary:
    print(key, ":", dictionary[key])
```
