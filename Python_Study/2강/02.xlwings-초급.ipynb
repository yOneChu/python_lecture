## 앱 실행, 워크북 만들기

```python
import xlwings as xw

## 엑셀 앱 만들기
app = xw.App(add_book=False)
```

```python
# 워크북 만들기
wb = app.books.add()
```

```python
# 워크시트 선택
ws = wb.sheets['sheet1']
ws
```

```python
# 시트 이름 변경
ws.name = "수배로직설계팀"
```

```python
# 저장
wb.save("excelTest.xlsx")
```

```python
# 엑셀 앱 닫기
app.quit()
```

## 예제 2.

```python
# 엑셀 워크북 불러오기
app = xw.App(add_book=False)
wb = app.books.open("excelTest.xlsx")
```

```python
# 새로운 시트 생성
wb.sheets.add("수배팀")
```

```python
# 시트 삭제
wb.sheets["수배팀"].delete()
```

## 셀 데이터 입력

```python
## 시트 제어권 가져오기
ws = wb.sheets["수배로직설계팀"]

# 값 입력
ws.range('A1').value = '이름'
ws.range('B1').value = '나이'
```

```python
## 여러 데이터 입력
ws.range('A2').value = ['김영환', 5]
```

```python
## 2차워 데이터 넣기
ws.range('A3').value = [['선우', 20], ['대환', 10]]
```

```python
## 세로로 넣기
ws.range('B5').options(transpose=True).value = [10.5, 100]
```

```python
## 셀 여러개 접근하기
ws.range('A1:B4').value
```

```python
## 합계 구하기
ws.range('A5').value = "합계"
ws.range('B5').value = "=sum(B2:B4)"
```

```python
## 값 복사
ws.range('B2:B6').copy()
```

```python
## 값 붙여넣기
ws.range('C2').paste()
ws.range('D2').paste()
```

```python
## 셀 확장해서 선택
ws.range("A1").expand("down").value
```

```python
## 셀 확장해서 선택 (오른쪽)
ws.range("A1").expand("right").value
```

```python
## 테이블로 가져오기
ws.range('A1').expand("table").value
```

```python
# 폰트 변경
ws.range('A1:B5').font.size = 13

ws.range('A1:B5').font.bold = True

ws.range('A1:B5').color = (255, 125, 0)
```

```python
## 저장하고 끝내기
wb.save()
app.quit()
```
